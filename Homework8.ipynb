```python
def is_leap_year(year):
    return (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)

def get_next_date(date_str):
    parts = date_str.split('/')
    if len(parts) != 3:
        return "INVALID"
    
    day = int(parts[0])
    month = int(parts[1])
    year = int(parts[2])
    
    if year < 1 or month < 1 or month > 12 or day < 1:
        return "INVALID"
        
    days_in_months = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if is_leap_year(year):
        days_in_months[2] = 29
        
    if day > days_in_months[month]:
        return "INVALID"
        
    day += 1
    if day > days_in_months[month]:
        day = 1
        month += 1
        if month > 12:
            month = 1
            year += 1
            
    return f"{day:02d}/{month:02d}/{year:04d}"

input_date = input()
print(get_next_date(input_date))
```

### W8A7	
Yêu cầu
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:

Rút gọn phân số trong __init__ (quy ước mẫu > 0).

Các phương thức add/sub/mul/div trả về phân số đã rút gọn.

__str__() in dạng a/b.

Input
a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})

Output
Phân số kết quả đã rút gọn


```python
def calculate_gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return abs(a)

class Fraction:
    def __init__(self, num, den):
        gcd = calculate_gcd(num, den)
        self.num = num // gcd
        self.den = den // gcd
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den
            
    def add(self, other):
        return Fraction(self.num * other.den + other.num * self.den, self.den * other.den)
        
    def sub(self, other):
        return Fraction(self.num * other.den - other.num * self.den, self.den * other.den)
        
    def mul(self, other):
        return Fraction(self.num * other.num, self.den * other.den)
        
    def div(self, other):
        return Fraction(self.num * other.den, self.den * other.num)
        
    def __str__(self):
        if self.den == 1:
            return f"{self.num}"
        return f"{self.num}/{self.den}"

def run_fraction_test():
    parts = input().split()
    a, b = int(parts[0]), int(parts[1])
    op = parts[2]
    c, d = int(parts[3]), int(parts[4])
    
    f1 = Fraction(a, b)
    f2 = Fraction(c, d)
    
    if op == '+':
        print(f1.add(f2))
    elif op == '-':
        print(f1.sub(f2))
    elif op == '*':
        print(f1.mul(f2))
    elif op == '/':
        print(f1.div(f2))

run_fraction_test()
```

rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8, "Average" nếu 5–<6.5, ngược lại "Poor".

Input

Dòng 2: n
n dòng tiếp: subject score (score là số thực 0–10)

Output

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []
        
    def add_score(self, score):
        self.scores += [score]
        
    def avg(self):
        if len(self.scores) == 0:
            return 0.0
        total = 0
        for s in self.scores:
            total += s
        return total / len(self.scores)
        
    def rank(self):
        average_score = self.avg()
        if average_score >= 8:
            return "Excellent"
        elif average_score >= 6.5:
            return "Good"
        elif average_score >= 5:
            return "Average"
        else:
            return "Poor"

def run_student_test():
    name = input()
    num_subjects = int(input())
    student = Student(name)
    
    for _ in range(num_subjects):
        subject_line = input().split()
        score = float(subject_line[1])
        student.add_score(score)
        
    print(f"{student.name} {student.avg():.2f} {student.rank()}")

run_student_test()
```

### W8A11	
Bạn được cho bốn điểm A, B, C và D trong hệ tọa độ Đề-các 3 chiều.
Nhiệm vụ của bạn là tính và in ra góc giữa hai mặt phẳng tạo bởi các điểm:
mặt phẳng (A, B, C)
mặt phẳng (B, C, D)
Góc cần in ra được ký hiệu là PHI, và phải dùng độ (degrees), không dùng radian.
Dạng Input:
Một dòng input chứa 3 số thực, ngăn cách bằng dấu cách, biểu diễn:
tọa độ X tọa độ Y tọa độ Z của một điểm.
Dạng Output: 
In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập phân.


```python
import math

def get_cross_product(v1, v2):
    return [
        v1[1] * v2[2] - v1[2] * v2[1],
        v1[2] * v2[0] - v1[0] * v2[2],
        v1[0] * v2[1] - v1[1] * v2[0]
    ]

def get_vector(p1, p2):
    return [p2[0] - p1[0], p2[1] - p1[1], p2[2] - p1[2]]

def get_dot_product(v1, v2):
    return v1[0] * v2[0] + v1[1] * v2[1] + v1[2] * v2[2]

def get_magnitude(v):
    return (v[0]**2 + v[1]**2 + v[2]**2)**0.5

def calculate_plane_angle():
    a = list(map(float, input().split()))
    b = list(map(float, input().split()))
    c = list(map(float, input().split()))
    d = list(map(float, input().split()))
    
    ab = get_vector(a, b)
    ac = get_vector(a, c)
    bc = get_vector(b, c)
    bd = get_vector(b, d)
    
    normal1 = get_cross_product(ab, ac)
    normal2 = get_cross_product(bc, bd)
    
    dot_val = get_dot_product(normal1, normal2)
    mag1 = get_magnitude(normal1)
    mag2 = get_magnitude(normal2)
    
    cos_theta = abs(dot_val) / (mag1 * mag2)
    cos_theta = min(1.0, max(-1.0, cos_theta))
    
    angle_rad = math.acos(cos_theta)
    angle_deg = math.degrees(angle_rad)
    print(f"{angle_deg:.2f}")

calculate_plane_angle()
```

Khai báo kiểu dữ liệu biểu diễn số phức \(ComplexNumber\) bao gồm \(2\) biến \(realPart\) và \(imaginaryPart\) kiểu \(double\) biểu diễn phần thực và phần ảo của một số phức.

Viết hàm khởi tạo ComplexNumber(double initRealPart, double initImaginaryPart) để khởi tạo các giá trị phần thực và phần ảo của một số phức và hàm ComplexNumber multiply(complexNumber other) trả về tích của số phức hiện tại và một số phức khác.

Đầu vào

Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo của một số phức. Các số trong đầu vào có giá trị tuyệt đối không vượt quá \(100\).

Đầu ra

In ra màn hình hai số biểu diễn phần thực và phần ảo của tích hai số phức.


```python
class ComplexNumber:
    def __init__(self, initRealPart, initImaginaryPart):
        self.realPart = initRealPart
        self.imaginaryPart = initImaginaryPart
        
    def multiply(self, other):
        res_real = self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart
        res_imag = self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart
        return ComplexNumber(res_real, res_imag)

def run_complex_test():
    r1, i1 = map(float, input().split())
    r2, i2 = map(float, input().split())
    
    c1 = ComplexNumber(r1, i1)
    c2 = ComplexNumber(r2, i2)
    result = c1.multiply(c2)
    
    real_str = f"{int(result.realPart)}" if result.realPart.is_integer() else f"{result.realPart}"
    imag_str = f"{int(result.imaginaryPart)}" if result.imaginaryPart.is_integer() else f"{result.imaginaryPart}"
    print(f"{real_str} {imag_str}")

run_complex_test()
```

### W8A13	
[Triangle]

Khai báo cấu trúc dữ liệu tam giác \(Triangle\) chứa \(3\) biến \(egde1,\,edge2,\,edge3\) kiểu \(int\) là độ dài \(3\) cạnh của tam giác đó.

Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm double getArea() trả về diện tích của tam giác đó (nếu tam giác không hợp lệ, trả về \(-1\)).

Hoàn thiện chương trình dưới đây nhập vào \(3\) cạnh của một tam giác và in ra diện tích của tam giác đó.

Đầu vào

Một dòng duy nhất từ bàn phím chứa \(3\) số nguyên là độ dài \(3\) cạnh của tam giác

Đầu ra

In ra màn hình diện tích tam giác (làm tròn đến\(2\) chữ số thập phân sau dấu phẩy). In ra \(invalid\) nếu \(3\) số nhập vào không phải là \(3\) cạnh của một tam giác


```python
class Triangle:
    def __init__(self, initEdge1, initEdge2, initEdge3):
        self.edge1 = initEdge1
        self.edge2 = initEdge2
        self.edge3 = initEdge3
        
    def getArea(self):
        a, b, c = self.edge1, self.edge2, self.edge3
        if a + b > c and a + c > b and b + c > a:
            s = (a + b + c) / 2
            area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
            return area
        return -1

def run_triangle_test():
    e1, e2, e3 = map(int, input().split())
    triangle = Triangle(e1, e2, e3)
    result = triangle.getArea()
    if result == -1:
        print("invalid")
    else:
        print(f"{result:.2f}")

run_triangle_test()
```

Xây dựng lớp Printer với phương thức print_value(x).
Xây dựng lớp BinaryPrinter kế thừa từ Printer, ghi đè print_value(x) để in biểu diễn nhị phân của
số nguyên dương x.Input
Một số nguyên dương n.
Output
In ra dạng nhị phân của n.

```python
class Printer:
    def print_value(self, x):
        print(x)

class BinaryPrinter(Printer):
    def print_value(self, x):
        if x == 0:
            print(0)
            return
        binary_str = ""
        while x > 0:
            binary_str = str(x % 2) + binary_str
            x //= 2
        print(binary_str)

def run_printer_test():
    n = int(input())
    printer = BinaryPrinter()
    printer.print_value(n)

run_printer_test()
```

```python
class Counter:
    def __init__(self):
        self.count = 0
        
    def inc(self):
        self.count += 1
        
    def dec(self):
        self.count -= 1
        
    def value(self):
        return self.count

def run_counter_test():
    n = int(input())
    operations = input().split()
    counter = Counter()
    
    for i in range(n):
        if operations[i] == "INC":
            counter.inc()
        elif operations[i] == "DEC":
            counter.dec()
            
    print(counter.value())

run_counter_test()
```

### W8A30 
Yêu cầu
Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x),
pop(), top().
Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ.
Input
Dòng 1: n
n dòng tiếp theo là một trong ba lệnh:
PUSH x
POP
TOP
Output
Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng.

```python
class Stack:
    def __init__(self):
        self.items = []
        
    def push(self, x):
        self.items += [x]
        
    def pop(self):
        if len(self.items) > 0:
            popped_item = self.items[len(self.items) - 1]
            self.items = self.items[:len(self.items) - 1]
            return popped_item
            
    def top(self):
        if len(self.items) > 0:
            return self.items[len(self.items) - 1]

def run_stack_test():
    n = int(input())
    stack = Stack()
    
    for _ in range(n):
        command_line = input().split()
        cmd = command_line[0]
        
        if cmd == "PUSH":
            val = int(command_line[1])
            stack.push(val)
        elif cmd == "POP":
            stack.pop()
        elif cmd == "TOP":
            print(stack.top())

run_stack_test()
```
